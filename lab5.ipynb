```python
num = 10
num1, num2 = 0, 1
for i in range(num):
    num3 = num1 + num2
    num1, num2= num2, num3
    print(f"The {num}th Fibonacci number is: {num3}")
```

```python
n = 10
a = 0
b = 1
for i in range(n):
    print(a, end=' ')
    a, b = b, a + b
```

```python
n = 10
a = 0
b = 1
while True:
    print(a, end=' ')
    a, b = b, a + b
    if b >= n:
        break
```

```python
def fib_series(n):
    num1 = 0
    num2 = 1
    while True:
        print(num1, end=' ')
        if num2 > n:
            break
        num1, num2 = num2, num1 + num2
        
fib_series(10)
```
